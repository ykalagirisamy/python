```python
BMI = float(input("Enter your BMI index: "))

if BMI < 18.5:
    print("Under weight")
elif BMI < 25.0:
    print("Healthy weight")
elif BMI < 30.0:
    print("Over weight")
elif BMI < 35.0:
    print("Obese class 1")
elif BMI < 40.0:
    print("Obese class 2")
else:
    print("Obese class 3")
```
